### VectorDB Instance

````python
# TODO: Instantiate your ChromaDB Client
# Choose any path you want
chroma_client = chromadb.PersistentClient(path="chromadb") 
````

### Collection

````python
# TODO: Pick one embedding function
# If picking something different than openai, 
# make sure you use the same when loading it
embedding_fn = embedding_functions.OpenAIEmbeddingFunction()
````

````python
# TODO: Create a collection
# Choose any name you want
collection = chroma_client.create_collection(
    name="udaplay",
    embedding_function=embedding_fn
)
````

### Add documents

````python
# Make sure you have a directory "project/starter/games"
data_dir = "games"

for file_name in sorted(os.listdir(data_dir)):
    if not file_name.endswith(".json"):
        continue

    file_path = os.path.join(data_dir, file_name)
    with open(file_path, "r", encoding="utf-8") as f:
        game = json.load(f)

    content = f"[{game['Platform']}] {game['Name']} ({game['YearOfRelease']}) - {game['Description']}"

    doc_id = os.path.splitext(file_name)[0]

    collection.add(
        ids=[doc_id],
        documents=[content],
        metadatas=[game],
    )
````

````python
# Test semantic search
results = collection.query(query_texts=["racing game on PlayStation"], n_results=3)
for i, doc in enumerate(results["documents"][0]):
    print(f"{i+1}. {doc}")


custom_games = [
    {
        "Name": "Balatro",
        "Platform": "PC",
        "Genre": "Roguelike",
        "Publisher": "LocalThunk",
        "Description": "A poker-inspired roguelike deck-building game.",
        "YearOfRelease": 2024
    }
]
for game in custom_games:
    content = f"[{game['Platform']}] {game['Name']} ({game['YearOfRelease']}) - {game['Description']}"
    collection.add(ids=[game['Name']], documents=[content], metadatas=[game])   
````
